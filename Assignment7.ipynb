```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load Titanic Dataset
df = pd.read_csv(r"C:\Users\MANAN\Downloads\Titanic-Dataset.csv")

# Display Dataset
print("Titanic Dataset:")
print(df.head())

# 1. Line Plot
plt.figure(figsize=(8, 5))

plt.plot(
    df["PassengerId"][:30],
    df["Fare"][:30],
    marker="o"
)

plt.title("Passenger Fare - Line Plot")
plt.xlabel("Passenger ID")
plt.ylabel("Fare")
plt.grid(True)

plt.show()


# 2. Bar Plot
plt.figure(figsize=(8, 5))

avg_fare = df.groupby("Pclass")["Fare"].mean()

plt.bar(
    avg_fare.index.astype(str),
    avg_fare.values
)

plt.title("Average Fare by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")

plt.show()


# 3. Scatter Plot
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Age",
    y="Fare",
    hue="Survived"
)

plt.title("Age vs Fare - Scatter Plot")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()


# 4. Correlation Matrix and Heatmap
correlation = df[
    ["Age", "Fare", "Pclass", "Survived"]
].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Titanic Dataset - Correlation Heatmap")

plt.show()
```
