```python
#To check if given triangle is right angle triangle while using function

def check(a,b,c) : 
    sum_a = a*a + b*b
    sum_b = c**2 + b*b
    sum_c = a*a + c**2
    if sum_a  == c**2 :
        print("It is right angle triangle!")
    elif sum_b == a**2 :
        print("It is right angle triangle!")
    elif sum_c == b**2 :
        print("It is right angle triangle!")
    else:
        print("It is not right angle triangle!")


num_1 = int(input("a:"))
num_2 = int(input("b:"))
num_3 = int(input("c:"))
check (num_1,num_2,num_3)
```
