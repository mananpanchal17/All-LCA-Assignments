```python
# Odd or Even
num = int(input("Enter a number: "))
if num % 2 == 0:
    print(" It is an Even number.")
else:
    print("It is an Odd number.")
```

```python
#Largest Three Numbers
n1=int(input("Enter First Number"))
n2=int(input("Enter Second Number"))
n3=int(input("Enter Third Number"))
if (n1>n2 and n1>n3):
  print("The Largest Number is", n1)
if(n2>n1 and n2>n3):
  print("The Largest Number", n2)
else:
  print("The Largest Number", n3)
```

```python
"""
Program to create a Dictionary, Tuple, and list of Students
and perform Add, Delete and Update operations on Dictionary
Creating Students list
"""
Student_list=["Dosa","Batata vada","Pizza"]
Student_T=("Dosa","Batata vada","Pizza")
Student_d={
    1:"Dosa",
    2:"Batata vada",
    3:"Pizza"
}
print("Student List")
print(Student_list)

print("Student Tuple")
print(Student_T)

print("\nStudent Dictionary")
print(Student_d)
#add section
Student_d[4]="Pav Bhaji"
print("\nAfter Adding ")
print(Student_d)
#Update Operations
Student_d[2]="Pani Puri"
print("\nAfter Updating")
print(Student_d)
#Delete Operations
del Student_d[3]
print("\nAfter Deleting")
print(Student_d)
#Append
Student_list.append("shawarma")
print("\nAfter Appending")
print(Student_list)
```
