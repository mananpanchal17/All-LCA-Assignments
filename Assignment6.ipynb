```python
import numpy as np
import pandas as pd

# 1. Create a sample dataset with missing values and duplicates.
data = {
    "Name": ["Aarav", "Diya", "Rohan", "Sneha", "Diya", "Kabir", "Meera", None],
    "Marks": [85, 90, np.nan, 78, 90, 66, np.nan, 72],
    "City": ["Pune", "Mumbai", "Pune", None, "Mumbai", "Nashik", "Pune", "Nagpur"],
}

df = pd.DataFrame(data)

# Alternatively, load data from a file:
# df = pd.read_csv("students.csv")
# df = pd.read_excel("students.xls")


# 2. Explore the dataset.
print("First five rows:")
print(df.head())

print("\nDataset information:")
df.info()

print("\nDescriptive statistics:")
print(df.describe())

print("\nMissing values in each column:")
print(df.isnull().sum())


# 3. Handle missing values.

# Replace missing marks with the mean.
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# Replace missing cities with the most frequent city.
df["City"] = df["City"].fillna(df["City"].mode()[0])

# Remove rows where the student's name is missing.
df = df.dropna(subset=["Name"])


# 4. Remove duplicate rows.
print("\nNumber of duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()


# 5. Select and filter data using loc[].
print("\nName and Marks columns:")
print(df.loc[:, ["Name", "Marks"]])

print("\nStudents with marks greater than 75:")
print(df.loc[df["Marks"] > 75])

print("\nStudents from Pune:")
print(df.loc[df["City"] == "Pune", ["Name", "Marks"]])


# 6. Display and save the cleaned dataset.
print("\nCleaned DataFrame:")
print(df)

df.to_csv("cleaned_students.csv", index=False)

print("\nCleaned data saved to 'cleaned_students.csv'.")
```
